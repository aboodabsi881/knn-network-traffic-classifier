```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['normal.'] = le.fit_transform(df['normal.'])
df['tcp'] = le.fit_transform(df['tcp'])
df['http'] = le.fit_transform(df['http'])
df['SF'] = le.fit_transform(df['SF'])
```

```python
print(df.dtypes)
```

```python
X = df.drop('normal.', axis=1)
y= df['normal.']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)#train is the big data bigger than x_test so we need to do fit transform
X_test = sc.transform(X_test) # but here we need just write transform not fit_transform
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
KNNclassifier = KNeighborsClassifier(n_neighbors=5)
KNNclassifier.fit(X_train,y_train) #Suppervised this is becuse we take X_train, y_train
```

```python
y_test
```

```python
#predicting the test set result
y_pred = KNNclassifier.predict(X_test)
```
